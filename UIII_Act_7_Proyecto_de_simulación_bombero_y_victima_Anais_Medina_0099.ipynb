#### Problema bombero y victima

```python
# --- ASISTENTE DE RESCATE INTELIGENTE (BOMBERO IA) ---
# Simulación de toma de decisiones en emergencias para estudiantes.

opcion = ""

# Ciclo principal para mantener el menú activo (mientras no se elija salir)
while opcion != "4":
    print("\n--- MENÚ DE EMERGENCIA IA ---")
    print("1. Evaluar estado de la víctima")
    print("2. Evaluar riesgo del entorno (Humo y Fuego)")
    print("3. Calcular tiempo estimado de rescate")
    print("4. Salir del sistema")

    opcion = input("Seleccione una opción (1-4): ")

    # Opción 1: Evaluación de la víctima (Lógica de diagnóstico)
    if opcion == "1":
        print("\n[IA] Iniciando diagnóstico de víctima...")
        esta_consciente = input("¿La víctima responde a estímulos? (si/no): ").lower()
        respira = input("¿La víctima respira con dificultad? (si/no): ").lower()

        # Uso de operadores lógicos para determinar prioridad
        if esta_consciente == "si" and respira == "no":
            print("DIAGNÓSTICO: Prioridad Verde. Estabilizar y escoltar.")
        elif esta_consciente == "si" and respira == "si":
            print("DIAGNÓSTICO: Prioridad Amarilla. Requiere máscara de oxígeno.")
        elif esta_consciente == "no":
            print("DIAGNÓSTICO: ¡ALERTA ROJA! Extracción inmediata y primeros auxilios.")
        else:
            print("DIAGNÓSTICO: Datos insuficientes. Repetir evaluación.")

    # Opción 2: Riesgo del entorno (Lógica de seguridad)
    elif opcion == "2":
        print("\n[IA] Evaluando sensores de entorno...")
        temperatura = float(input("Ingrese temperatura detectada (Celsius): "))
        visibilidad = int(input("Nivel de visibilidad (1-100%): "))

        # Evaluación de umbrales con operadores relacionales
        if temperatura > 100 or visibilidad < 20:
            print("RIESGO: EXTREMO. No ingresar sin equipo autónomo de respiración.")
        elif temperatura > 50 and temperatura <= 100:
            print("RIESGO: ALTO. Tiempo de permanencia limitado a 5 minutos.")
        else:
            print("RIESGO: MODERADO. Proceder con precaución.")

    # Opción 3: Cálculo aritmético simple
    elif opcion == "3":
        distancia = float(input("Distancia hasta la víctima (metros): "))
        obstaculos = input("¿Hay escombros bloqueando el camino? (si/no): ").lower()

        # Cálculo base
        tiempo = distancia / 2  # Asumimos 2 metros por segundo

        if obstaculos == "si":
            tiempo = tiempo + 30  # Sumar 30 segundos por obstáculos
            print(f"TIEMPO ESTIMADO: {tiempo} segundos (incluye despeje de escombros).")
        else:
            print(f"TIEMPO ESTIMADO: {tiempo} segundos. Camino despejado.")

    # Opción de salida
    elif opcion == "4":
        print("Cerrando sistema de asistencia. ¡Manténgase a salvo, bombero!")

    # Manejo de opciones inválidas
    else:
        print("Opción no válida. Intente de nuevo.")

print("Sistema fuera de línea.")
print("Programa realizado por Anais Medina NL = 40")
```
