#### Problema bombero y extinguidor

```python
# --- SISTEMA IA: ASISTENTE DE EXTINCIÓN PARA BOMBEROS ---
# Simulación interactiva de toma de decisiones.

opcion = "0"

# Ciclo principal para mantener el menú activo hasta que el usuario decida salir [3]
while opcion != "4":
    print("\n========================================")
    print("   MENU DE ASISTENCIA TÉCNICA (IA)   ")
    print("========================================")
    print("1. Identificar extintor por tipo de fuego")
    print("2. Verificar estado del extintor")
    print("3. Protocolo de ataque de emergencia")
    print("4. Apagar sistema")

    opcion = input("\nSeleccione una opción (1-4): ")

    # Opción 1: Lógica para seleccionar el agente extintor correcto [4, 5]
    if opcion == "1":
        print("\n[IA] Analizando materiales combustibles...")
        print("Tipos: 1. Madera/Papel, 2. Líquidos/Grasas, 3. Eléctrico")
        tipo = input("Ingrese el número del tipo de material: ")

        if tipo == "1":
            print("RECOMENDACIÓN: Use Extintor de Agua o Polvo Químico Clase A.")
        elif tipo == "2":
            print("RECOMENDACIÓN: Use Extintor de Espuma o PQS Clase B.")
        elif tipo == "3":
            # Nota de seguridad: El agua conduce electricidad [18.1]
            print("RECOMENDACIÓN CRÍTICA: Use CO2 (Dióxido de Carbono). ¡No use agua!")
        else:
            print("ALERTA: Tipo de material desconocido. Proceda con precaución extrema.")

    # Opción 2: Evaluación de sensores del equipo (Aritmética y Relacional) [6, 7]
    elif opcion == "2":
        print("\n[IA] Verificando presión y carga...")
        presion = float(input("Ingrese la presión del manómetro (en PSI): "))
        peso = float(input("Ingrese el peso actual del tanque (kg): "))

        # Uso de operadores lógicos para determinar si el equipo es confiable [8, 9]
        if presion >= 100 and presion <= 150:
            estado_presion = "ÓPTIMA"
        else:
            estado_presion = "FUERA DE RANGO (Peligro)"

        if peso < 5.0:
            print(f"ESTADO: {estado_presion}. Carga BAJA. Reemplazar extintor pronto.")
        else:
            print(f"ESTADO: {estado_presion}. Carga SUFICIENTE.")

    # Opción 3: Toma de decisiones compleja (Operadores lógicos combinados) [10, 11]
    elif opcion == "3":
        print("\n[IA] Evaluación de riesgo de ataque...")
        distancia = float(input("¿A qué distancia está el fuego (metros)?: "))
        viento_fuerte = input("¿Hay viento fuerte en contra? (si/no): ").lower()
        humo_denso = input("¿La visibilidad es nula por humo? (si/no): ").lower()

        # Decisión basada en múltiples factores de riesgo
        if (distancia < 2) or (viento_fuerte == "si" and humo_denso == "si"):
            print("IA SUGIERE: ¡RETROCEDA! Riesgo de quemaduras o desorientación alto.")
        elif viento_fuerte == "no" and distancia >= 3:
            print("IA SUGIERE: Posición segura. Inicie descarga en la base de las llamas.")
        else:
            print("IA SUGIERE: Ataque con precaución. Use equipo de respiración.")

    # Opción de salida
    elif opcion == "4":
        print("\nCerrando sistema de asistencia. ¡Manténgase seguro, bombero!")

    # Manejo de entradas no válidas en el menú [12]
    else:
        print("\n[Error] Opción no válida. Intente de nuevo.")

print("\n--- Sistema fuera de línea ---")
print("programa realizado por Anais Medina NL = 40")
```
